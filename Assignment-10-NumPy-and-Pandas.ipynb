```python
import numpy as np
import pandas as pd
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)

# Inspection and Properties:

print(temperatures_w1.shape)
print(temperatures_w1.dtype)
print(temperatures_w1.size)
print(temperatures_w1.ndim)

# Array Operations: 

temperatures_Fahrenheit = (temperatures_w1 * 9/5) + 32 
print(temperatures_Fahrenheit)

print(temperatures_w1.max())
print(temperatures_w1.min())
print(temperatures_w1.mean())


# Array Slicing and Indexing: 
print(temperatures_w1[0:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])


# 5. Create a 2D Array: 

temperatures = np.array([
     [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
     [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print(temperatures)

# Inspect and Slice the 2D Array: 

print(temperatures.shape)
print(temperatures.dtype)
print(temperatures.size)
print(temperatures.ndim)

print(temperatures[0])
print(temperatures[1])
print(temperatures[:, -2:])
```

```python
marks = pd.Series( [95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)


# Indexing and Slicing:

print(marks.iloc[0])
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
print(marks.iloc[2])
print(marks[marks > 90])


marks.loc['Rank1']= 100
print(marks)

marks= marks.drop('Rank5')
print(marks)

cgpa = marks / 10
print(cgpa)
```
